```python
# vectorisation tfidf 
tdidf_groupes = {}
for numero, liste in groupes20.items():
    tdidf_groupes[numero] = pretraitement.tf_idf_avec_son_dict(liste, vocabulaire)
```

```python
for i in range(20):
    print(len(tdidf_groupes[i]))
```

```python
# 20 groupes -> X
X = list()
Y = []
# je ne prends que 50% du jeu de données initiales 
for g , liste in tdidf_groupes.items():
    i = 0 
    for vecteur in liste:
        if i%2 == 0 :
            X.append(vecteur)
            Y.append(g)
        i += 1
```

```python
# Classifieur KNN 
sauvegarde = []
X = np.array(X)
Y = np.array(Y)
for k in range(1,30,5):
    knn = classif.ClassifierKNN(len(vocabulaire),k)
    perf, taux_moyen, taux_ecart = ev.validation_croisee(knn,(X,Y),10)
    sauvegarde.append((perf,taux_moyen, taux_ecart))
    # graphique l'axe x : k ,     l'axe Y : score de taux moyen et taux d'écart 
```

```python
taux_moyens
```

```python

# Récupération des données
k_values = list(range(1, 30, 5))
taux_moyens = [t[1] for t in sauvegarde]
taux_ecarts = [t[2] for t in sauvegarde]

# Création du graphique
plt.figure(figsize=(10, 6))
plt.errorbar(k_values, taux_moyens, yerr=taux_ecarts, fmt='-o', capsize=5, label="Taux moyen ± écart-type")

plt.title("Évolution du taux de bonne classification selon k (KNN)")
plt.xlabel("Valeur de k (nombre de voisins)")
plt.ylabel("Taux de bonne classification")
plt.xticks(k_values)
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Perceptron biais
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

perceptron = classif.ClassifierPerceptron(len(vocabulaire))
MOAA = classif.ClassifierMultiOAA(perceptron)
perf_p, taux_moyen_p, taux_ecart_p, Y_true, Y_pred = ev.validation_croisee(MOAA, (X,Y), 10)

```

```python
# code utilisé depuis chatGPT
# j'ai créer ma propre fonction matrice confusion 
# mais la lisibilité de la matrice confusion fait maison est difficile à regarder et à distinguer

from sklearn.metrics import confusion_matrix

def afficher_matrice_pourcentage(Y_true, Y_pred, class_labels=None):
    """
    Affiche une matrice de confusion avec les pourcentages.
    
    Paramètres :
    ------------
    Y_true : liste ou array des classes réelles
    Y_pred : liste ou array des classes prédites
    class_labels : noms des classes à afficher (optionnel)
    """
    cm = confusion_matrix(Y_true, Y_pred)
    cm_normalized = cm.astype("float") / cm.sum(axis=1)[:, np.newaxis]  # pourcentages ligne par ligne

    fig, ax = plt.subplots(figsize=(14, 12))
    im = ax.imshow(cm_normalized, interpolation="nearest", cmap=plt.cm.Blues)

    plt.colorbar(im)
    classes = class_labels if class_labels is not None else np.unique(Y_true)

    ax.set(
        xticks=np.arange(len(classes)),
        yticks=np.arange(len(classes)),
        xticklabels=classes,
        yticklabels=classes,
        ylabel="Classe réelle",
        xlabel="Classe prédite",
        title="Matrice de confusion (% par classe réelle)"
    )

    plt.xticks(rotation=45)

    # Étiquettes dans chaque case
    fmt = ".1%"  # format pourcentage
    thresh = cm_normalized.max() / 2.0
    for i in range(cm.shape[0]):
        for j in range(cm.shape[1]):
            ax.text(
                j, i,
                format(cm_normalized[i, j], fmt),
                ha="center", va="center",
                color="white" if cm_normalized[i, j] > thresh else "black"
            )

    plt.tight_layout()
    plt.show()
```

```python
afficher_matrice_pourcentage(Y_true, Y_pred, class_labels=range(20))
```
